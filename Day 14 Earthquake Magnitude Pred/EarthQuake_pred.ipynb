```python
import reverse_geocoder as rg

def get_country_rg(lat, lon):
    try:
        result = rg.search((lat, lon))  # tuple
        return result[0]['cc']  # country code
    except:
        return None
```

```python
coords = list(zip(df['latitude'], df['longitude']))

results = rg.search(coords)
```

```python
results
```

```python
df['Country_Code'] = [r['cc'] for r in results]
```

```python
import pycountry

def code_to_country(code):
    try:
        return pycountry.countries.get(alpha_2=code).name
    except:
        return None

df['Country'] = df['Country_Code'].apply(code_to_country)
```
